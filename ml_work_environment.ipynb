## ML Helper-Funktionen

**`prepare_ml_data(data, target, features)`** - Trennt Daten in Features (X) und Zielvariable (y)

**`split_data(X, y, test_size=0.2)`** - Teilt in Trainings- und Testdaten auf

**`train_model(X_train, y_train, model_type, ...)`** - Trainiert ein Modell
- `model_type='decision_tree'` - Entscheidungsbaum
- `model_type='random_forest'` - Mehrere Bäume
- `model_type='knn'` - K-Nearest Neighbors
- `model_type='svm'` - Support Vector Machine
- `model_type='gradient_boosting'` - Boosted Trees (fortgeschritten)

**`evaluate_model(model, X_test, y_test)`** - Bewertet das Modell

**`predict(model, data, features)`** - Macht Vorhersagen fuer neue Daten

**`plot_confusion_matrix(y_true, y_pred)`** - Zeigt welche Klassen verwechselt werden

**`plot_feature_importance(model, feature_names)`** - Zeigt welche Features wichtig sind

**`plot_decision_tree(model, feature_names)`** - Visualisiert den Entscheidungsbaum
